# Problem

If we list all the natural numbers below 10 that are multiples of 3 or 5, we get 3, 5, 6 and 9. The sum of these multiples is 23.

Find the sum of all the multiples of 3 or 5 below 1000.

## scratch work

A naive solution would simply go through all the numbers up to 1000 and check if each are 0 mod 3 or 0 mod 5, then sum all numbers that meet this condition.

A more efficient approach would be to recognise that the sum of multiples of a given number a, up to a maximum value m, form an arithmetic progression, and the sum of an arithmetic progression is

$$\frac{n(a_1 + a_n)}{2}$$

here $a_n$ is simply $m - (m \mod a) $ when $m \mod a \ne 0$, otherwise it is simply $m - a$, and $n$ is $\lfloor m /3 \rfloor$.

## Naive solution

```python
def sum_of_multiples_of_3_and_5(n):
    """
    sums the multiples of 3 and 5 below a given n.
    """
    return sum([x for x in range(n) if (x % 3 == 0) or (x % 5 == 0)])

assert sum_of_multiples_of_3_and_5(10) == 23

k = 1000
print(f"The sum of all the multiples of 3 or 5 below {k} is {sum_of_multiples_of_3_and_5(k)}")
```

## Solution

```python
def sum_multiples(a, m):
    """
    Returns of sum of all multiples of a, up to m.

    inputs:
    a - The number we wish to sum multiples of
    m - The maximum of the sequence
    """
    n = (m - 1) // a
    a_n = (m - a) if ( m % a == 0) else m - (m % a)

    return n * (a + a_n) // 2

assert sum_multiples(3, 10) + sum_multiples(5, 10) == 23

k = 1000
# multiples of 15 are  double counted so subtract to balance this out
res = sum_multiples(3, k) + sum_multiples(5, k) - sum_multiples(15, k)
print(f"The sum of all the multiples of 3 or 5 below {k} is {res}")
```
