# Problem #0

A number is a perfect square, or a square number, if it is the square of a positive integer.
For example,  is a square number because $5^2 = 5 \times 5 = 25$   ; it is also an odd square.

The first 5 square numbers are: $1, 4, 9, 16, 25$, and the sum of the odd squares is $1 + 9 + 25 = 35$.

Among the first **300 thousand** square numbers, what is the sum of all the odd squares?

## Scratch work

Given a sequence of $n$ numbers, we can find the sum off the odd numbers by starting at 1 and counting up by 2 (the naive solution) or by deriving/looking up a formula for the sum of odd squares.

The formula for the sum of odd squares can be shown to be $\frac{n(2n-1)(2n+1)}{3} $

## Naive solution

```python
x = np.arange(1, 300000, 2)
squares = x * x
print("The sum of the odd squares up to 300,000 is ", np.sum(squares))
```

## Solution

```python
def sum_of_odd_squares(n):
    """
    returns the sum of the first n odd squares
    inputs:
    n - The number of odd squares to sum
    """
    return (4 * n ** 3 - n) // 3

# half of the first 300,000 square numbers are odd, so divide by 2
n = 150000
print("The sum of the odd squares up to 300,000 is ", sum_of_odd_squares(n))
```
